### Advance Operation With Business Example


```python
import numpy as np
import matplotlib.pyplot as plt
```

```python

data = np.array([
    [101, 52000, 54800, 55900, 61000, 64000],   # Blue Tokai Coffee Roasters
    [102, 45000, 46200, 48100, 49500, 52000],   # Starbucks
    [103, 38000, 41000, 43000, 44500, 47000],   # Third Wave Coffee
    [104, 30000, 31500, 32500, 34000, 36000],   # Café Coffee Day
    [105, 27500, 28900, 30500, 32000, 33500],   # The Coffee Bean & Tea Leaf
])
print("data shape:", data.shape);
print("data from 0 to 3:\n", data[:3])
print("all rows, first column:\n", data[:, 0:4])
print(" all column and forst 3 rows:\n", data[0:3, :])
print("total sales for all cafes:\n", data[:, 1:].sum(axis=1))
a=np.sum(data[:, 1:], axis=1)
print(a)
min_sales=np.min(data[:,1:],axis=1)
max_sales=np.min(data[:,1:],axis=0)
print("min sales:", min_sales)
print("max sales:", max_sales)
```

```python
cumsum= np.cumsum(data[:, 1:], axis=1)
print("Cumulative sum:\n", cumsum)
plt.figure(figsize=(10, 6))
plt.plot(np.mean(cumsum,axis=0))
plt.title("Cumulative Sales Over Time")
plt.xlabel("Months")
plt.ylabel("Cumulative Sales")
plt.grid()
plt.show()
```

### Vectors


```python
vector1=np.array([1,2,3,4,5])
vector2=np.array([6,7,8,9,10])
print("Vector 1:", vector1)
print("Vector 2:", vector2)
print("Element-wise addition:", vector1 + vector2)
print("Element-wise multiplication:", vector1 * vector2)    
print("Element-wise division:", vector1 / vector2)
print("Element-wise subtraction:", vector1 - vector2)
print("Element-wise exponentiation:", vector1 ** 2)
print("Element-wise square root:", np.sqrt(vector1))
print("Element-wise logarithm:", np.log(vector1))
print("Element-wise sine:", np.sin(vector1))
print("Element-wise cosine:", np.cos(vector1))      
dot_product = np.dot(vector1, vector2)
print("Dot product:", dot_product)
```
